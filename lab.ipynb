```python
from scipy.stats import norm

x1 = df['Daily_Footfall']
x2 = df['Discount']

mean1, mean2 = np.mean(x1), np.mean(x2)
std1, std2 = np.std(x1), np.std(x2)
n1, n2 = len(x1), len(x2)

z = (mean1 - mean2) / np.sqrt((std1**2/n1) + (std2**2/n2))
p = 2 * (1 - norm.cdf(abs(z)))

print("Z-Test:", z, p)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

# Handle missing values
df.fillna(df.mean(numeric_only=True), inplace=True)

# Features and target
X = df[['Daily_Footfall','Discount']]
y = df['Items_Sold']

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Model
model = LinearRegression()
model.fit(X_train, y_train)

# Prediction
new_data = pd.DataFrame([[150,20]], columns=['Daily_Footfall','Discount'])
pred = model.predict(new_data)

print("Prediction:", pred)
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

df['Sales_Category'] = df['Items_Sold'].apply(lambda x: 1 if x>40 else 0)

X = df[['Daily_Footfall','Discount']]
y = df['Sales_Category']

model = LogisticRegression()
model.fit(X,y)

# Correct prediction input
new_data = pd.DataFrame([[160,15]], columns=['Daily_Footfall','Discount'])

print("Logistic Prediction:", model.predict(new_data))
```
